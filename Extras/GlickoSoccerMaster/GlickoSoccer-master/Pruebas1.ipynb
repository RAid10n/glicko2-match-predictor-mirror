```python
glicko = Glicko2()
teamA = Rating(1500, 200, 0.06)
teamB = Rating(1400, 200, 0.06)

print("Probabilidades:", glicko.probabilities(teamA, teamB, advantage=0, skellam_draw_probability=0.1))
```

```python
newA, newB = glicko.rate(teamA, teamB, advantage=0, outcome='H', skellam_draw_probability=0.1)
print("Nuevo rating A:", newA)
print("Nuevo rating B:", newB)
```
